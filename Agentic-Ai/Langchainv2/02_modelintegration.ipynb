```python
#Model Inetgration with OPENAI
import os
from dotenv import load_dotenv
load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
import langchain
from langchain.chat_models import  init_chat_model
model = init_chat_model('gpt-4.1')
```

```python
##invoke the model
response = model.invoke("Hey, How are you?")
response.content
```

## Streaming and Batch 
* Streaming allows the model to stream their output content while it is being generated.
* Batch is a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel.


```python
for chunk in model.stream("write me a 100 words poem on Badminton"):
    print(chunk.text, end="", flush=True)
```

```python
responses = model.batch([
    'Who is the greatest player in Badminton?',
    'Who is the greatest player in Cricket?',
    'Who is the greatest player in Football?'

])
for response in responses:
    print(response.content)
```

```python
model.batch([
    'Who is the greatest player in Badminton?',
    'Who is the greatest player in Cricket?',
    'Who is the greatest player in Football?'
    ''
],
config ={
    'max_concurrency': 4 # limit to 4 parallel calls
}
)
```
