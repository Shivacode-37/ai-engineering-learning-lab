##Runnables 
* Runnable is anything that can take input and produce an output

```python
# How we create chain and classes Without runnable
import random

class F_llm:
    def __init__(self):
        print('LLM is working')
    def predict(self, prompt):

        response_generator_list=[
            'RCB become the defending champion',
            'Gojo is the storngest sorcerer',
            'Ai will take those people job, who is not ready to change'
        ]
        return {'response':random.choice(response_generator_list)}
```

```python
class F_prompttemplate:
    def __init__(self, template, input_variables):
        self.template = template
        self.input_variables = input_variables
    def format(self,input_dict):
        return self.template.format(**input_dict)
```

```python
template = F_prompttemplate(
    template= 'Give me the {length} response list information{topic}',
    input_variables=['length','topic']
)
```

```python
prompt = template.format({'length':'short', 'topic':'Ai'})
```

```python
class F_llmchain:
    def __init__(self, llm, prompt):
        self.llm = llm
        self.prompt = prompt

    def run(self, input_dict):
        final_prompt = self.prompt.format(input_dict)
        result = self.llm.predict(final_prompt)

        return result['response']
```

```python
template = F_prompttemplate(
    template= 'Give me the {length} response list information{topic}',
    input_variables=['length','topic']
)
```

```python
llm = F_llm()
```

```python
llm.predict(prompt)
```

```python
chain = F_llmchain(llm,template)
```

```python
chain.run({'length':'medium','topic':'Ai'})
```
