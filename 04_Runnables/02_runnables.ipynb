```python
# How to use chains with runnables
from abc import ABC, abstractmethod

class Runnable(ABC):

    @abstractmethod
    def invoke(input_data):
        pass
```

```python
import random

class R_llm(Runnable):
    def __init__(self):
        print('LLM is working')
    def invoke(self, prompt):
        response_generator_list = [
            'RCB won the 2026 IPL',
            'Asts become the magic Emperor',
            'With help Ai you can change your imagination into reality'
        ]
        return {'response':random.choice(response_generator_list)}

    def predict(self, prompt):

        response_generator_list=[
            'RCB become the defending champion',
            'Gojo is the storngest sorcerer',
            'Ai will take those people job, who is not ready to change'
        ]
        return {'response':random.choice(response_generator_list)}
```

```python
class R_promptTemplate(Runnable):
    def __init__(self, template, input_variables):
        self.template = template
        self.input_variables = input_variables

    def invoke(self, input_dict):
        return self.template.format(**input_dict)

    def format(self, input_dict):
        return self.template.format(**input_dict)
```

```python
class Rstroutputparser(Runnable):
    def __init__(self):
        pass
    def invoke(self, input_data):
        return input_data['response']
```

```python
class RunnableConnector(Runnable):
    def __init__(self,runnable_list):
        self.runnable_list = runnable_list

    def invoke(self,input_data):
        for runnable in self.runnable_list:
            input_data = runnable.invoke(input_data)
        return input_data

```

```python
template= R_promptTemplate(
    template ='Give me the {details} of the {topic}',
    input_variables= ['details','topic']
)
```

```python
llm = R_llm()
```

```python
parser = Rstroutputparser()
```

```python
chain = RunnableConnector([template, llm, parser])
chain.invoke({'details':'line','topic':'ai'})
```
